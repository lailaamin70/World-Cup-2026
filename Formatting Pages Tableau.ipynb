```python
import pandas as pd

# Age profile
players = pd.read_csv("C:/Users/laila/Documents/Football Project/player_summary_top20.csv")

age_long = players[['Nation','young_count','peak_count','aging_count']].melt(
    id_vars='Nation',
    var_name='Age Group',
    value_name='Count'
)
age_long['Age Group'] = age_long['Age Group'].replace({
    'young_count' : 'Under 24',
    'peak_count'  : '24-29 Peak',
    'aging_count' : '30+ Aging'
})
age_long.to_csv("C:/Users/laila/Documents/Football Project/age_profile.csv", index=False)

# Skills long format
skills_long = players[['Nation','PAC','SHO','PAS','DRI','DEF','PHY']].melt(
    id_vars='Nation',
    var_name='Skill',
    value_name='Rating'
)
skills_long.to_csv("C:/Users/laila/Documents/Football Project/skills_long.csv", index=False)

print("Both files saved!")
```

```python
import pandas as pd

mc = pd.read_csv("C:/Users/laila/Documents/Football Project/monte_carlo_results_2026.csv")

# Filter top 15 by win probability
top15 = mc.nlargest(15, 'win_pct')

# Reshape to long format
stage_long = top15[['TEAM', 'group_adv_pct', 'r16_pct', 
                     'semi_pct', 'final_pct', 'win_pct']].melt(
    id_vars='TEAM',
    var_name='Stage',
    value_name='Probability'
)

stage_long['Stage'] = stage_long['Stage'].replace({
    'group_adv_pct' : '1. Group Stage',
    'r16_pct'       : '2. Round of 16',
    'semi_pct'      : '3. Semi Final',
    'final_pct'     : '4. Final',
    'win_pct'       : '5. Winner'
})

stage_long.to_csv(
    "C:/Users/laila/Documents/Football Project/stage_progression.csv",
    index=False
)
print("Saved!")
```
