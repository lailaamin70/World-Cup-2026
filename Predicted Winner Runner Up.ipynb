```python
import pandas as pd

# Define the 2026 groups
groups = {
    'A': ['Mexico', 'South Africa', 'South Korea', 'Czech Republic'],
    'B': ['Canada', 'Bosnia and Herzegovina', 'Qatar', 'Switzerland'],
    'C': ['Brazil', 'Morocco', 'Haiti', 'Scotland'],
    'D': ['United States', 'Paraguay', 'Australia', 'Turkey'],
    'E': ['Germany', 'Curacao', 'Ivory Coast', 'Ecuador'],
    'F': ['Netherlands', 'Japan', 'Sweden', 'Tunisia'],
    'G': ['Belgium', 'Egypt', 'Iran', 'New Zealand'],
    'H': ['Spain', 'Cape Verde', 'Saudi Arabia', 'Uruguay'],
    'I': ['France', 'Senegal', 'Iraq', 'Norway'],
    'J': ['Argentina', 'Algeria', 'Austria', 'Jordan'],
    'K': ['Portugal', 'DR Congo', 'Uzbekistan', 'Colombia'],
    'L': ['England', 'Croatia', 'Ghana', 'Panama']
}

third_place_advances = ['A', 'B', 'D', 'E', 'F', 'G', 'J', 'K']


# Load composite scores
wc = pd.read_csv("C:/Users/laila/Documents/Football Project/wc2026_all48_summary.csv")

# Build group predictions table
rows = []
for grp, teams in groups.items():
    grp_letter = grp  
    grp_data = wc[wc['TEAM'].isin(teams)].copy()
    grp_data = grp_data.sort_values('composite_score', ascending=False)
    
    for i, (_, row) in enumerate(grp_data.iterrows()):
        if i == 0:
            status = '1st — Advances'
        elif i == 1:
            status = '2nd — Advances'
        elif i == 2:
            if grp_letter in third_place_advances:
                status = '3rd — Advances'
            else:
                status = '3rd — Eliminated'
        else:
            status = '4th — Eliminated'

        rows.append({
            'Group'           : f'Group {grp}',
            'Team'            : row['TEAM'],
            'Predicted Finish': i + 1,
            'Composite Score' : round(row['composite_score'], 3),
            'Confederation'   : row['confederation'],
            'Status'          : status
        })

group_predictions = pd.DataFrame(rows)

print(group_predictions[['Group', 'Team', 'Predicted Finish', 
                          'Status', 'Composite Score']].to_string(index=False))

group_predictions.to_csv(
    "C:/Users/laila/Documents/Football Project/group_predictions.csv",
    index=False
)
print("\nSaved!")
```
